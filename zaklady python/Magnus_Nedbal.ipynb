1. Pomocí funkce `datetime.strptime()` převeďte následující řetězce na objekt třídy `datetime`. (2b) 
* "1.1.2023 03h15m56s"
* "Feb 25 2023 4:20PM" 
* "20231208 14:56"
* "04/08/2023 03:10"
* "Wed 01/29/23 12:33PM"

```python
from datetime import datetime

s1 = "1.1.2023 03h15m56s"
dt1 = datetime.strptime(s1, "%d.%m.%Y %Hh%Mm%Ss")

s2 = "Feb 25 2023 4:20PM"
dt2 = datetime.strptime(s2, "%b %d %Y %I:%M%p")

s3 = "20231208 14:56"
dt3 = datetime.strptime(s3, "%Y%m%d %H:%M")

s4 = "04/08/2023 03:10"
dt4 = datetime.strptime(s4, "%m/%d/%Y %H:%M")

s5 = "Wed 01/29/23 12:33PM"
dt5 = datetime.strptime(s5, "%a %m/%d/%y %I:%M%p")

print(dt1)
print(dt2)
print(dt3)
print(dt4)
print(dt5)
```

2. Vytvořte program, který požádá uživatele o datum narození a řekne mu za kolik dní bude mít narozeniny. (2b)   

```python
from datetime import datetime, date

datum = input("Zadejte datum narození (DD.MM.YYYY): ")
narozeni = datetime.strptime(datum, "%d.%m.%Y").date()

dnes = date.today()

narozeniny_letos = date(dnes.year, narozeni.month, narozeni.day)

if narozeniny_letos < dnes:
    narozeniny_letos = date(dnes.year + 1, narozeni.month, narozeni.day)

rozdil = (narozeniny_letos - dnes).days

print(f"Zbývá {rozdil} dní do narozenin.")
```

3. Vytvořte simulátor pro hod dvěma kostkami. Pomocí simulace 1000 hodů dvěma kostkami vypočtěte četnost jednotlivých možných výsledků (součet 2 až součet 12) ve formě slovníku `dict`, kdy jednotlivé klíče budou výsledek hodu kostkami a hodnoty budou obsahovat počet výsledků stejné hodnoty z 1000 hodů. (2b)  

```python
import random

vysledky = {i: 0 for i in range(2, 13)}

for _ in range(1000):
    hod1 = random.randint(1, 6)
    hod2 = random.randint(1, 6)
    soucet = hod1 + hod2
    vysledky[soucet] += 1

print(vysledky)
```

4. Vytvořte funkci `fake_coin()`, která bude mít výstup buď "hlava" nebo "orel", kde pravděpodobnost, že padne hlava je ale 75%. (1b) 

```python
import random

def fake_coin():
    if random.random() < 0.75:
        return "hlava"
    else:
        return "orel"

pocet_hlava = 0
pocet_orel = 0

for _ in range(1000):
    vysledek = fake_coin()
    if vysledek == "hlava":
        pocet_hlava += 1
    else:
        pocet_orel += 1

print(f"Hlava padla: {pocet_hlava}x")
print(f"Orel padl: {pocet_orel}x")
print(f"Procento hlav: {pocet_hlava / 10}%")
```

5. Vytvořte program pro náhodné rozdání 32 karet mezi N hráčů (N je v rozsahu 2 až 16 ), tak aby měli všichni stejně karet. Zbytek karet zůstane v hromádce. (3b)

```python
import random

vals = ['7', '8', '9', '10', 'jack', 'queen', 'king', 'ace']
suits = ['spades', 'clubs', 'hearts', 'diamonds']
deck_of_cards = []

for val in vals:
    for suit in suits:
        deck_of_cards.append(val + "_" + suit)

def shuffle_cards(deck, N):
    if not (2 <= N <= 16):
        print("Chyba: Počet hráčů musí být v rozsahu 2 až 16.")
        return

    random.shuffle(deck)

    karet_na_hrace = len(deck) // N
    
    print(f"Počet hráčů: {N}, každý dostane {karet_na_hrace} karet.")

    index = 0
    for i in range(1, N + 1):
        ruka_hrace = deck[index : index + karet_na_hrace]
        print(f"Hráč {i}: {ruka_hrace}")
        
        index += karet_na_hrace

    zbytek = deck[index:]
    print(f"Zbytek v hromádce: {zbytek}")

shuffle_cards(deck_of_cards, 5)
```
